### Ollama on macOS

To run local LLMs on your Mac with Ollama:

1. Install Ollama:
   - Using Homebrew: `brew install ollama`
   - Or download the macOS installer from https://ollama.com/download/mac
2. Start Ollama:
   `ollama serve`
3. Pull a model:
   `ollama pull gemma3:latest`
   - You can also try other models like `llama3.1` or `mistral`.
4. Test it in the terminal:
   `ollama run gemma3:latest "Hello from Ollama!"`
5. Once the model is available locally, you can use it from Python with LangChain:

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(model="gemma3:latest")
response = llm.invoke("Hi! Can you help me build a simple app?")
print(response.content)
```

This repo uses the same pattern for local model inference without needing an external API key.
