```python
"""Streaming Responses Demo.

This notebook shows how to stream text from a local LLM (Ollama) and 
a cloud-hosted LLM (Groq) using LangChain. Streaming allows the model to
return tokens incrementally instead of waiting for the entire response to be
fully generated.
"""
```

```python
# Load environment variables from a local .env file if present.
# This is useful for API keys or other settings that should not be hardcoded.
from dotenv import load_dotenv
load_dotenv()
```

```python
# Create an Ollama-backed chat model configured for streaming responses.
# The model runs locally and returns tokens as they are generated.
from langchain_ollama import ChatOllama
llm = ChatOllama(model="gemma3:latest", streaming=True)
```

```python
# Request a streamed answer to a prompt about Agentic AI.
# The returned value is a stream object, not a final string yet.
response = llm.stream("Tell me about Agentic AI and how it is different from traditional AI models.")
```

```python
# Inspect the stream object to confirm that it is iterable and yields chunks.
response
```

```python
# Print each chunk in the stream along with a newline.
# This demonstrates the incremental nature of streamed generation.
for chunk in response:
    print(chunk, "\n")
```

```python
# Stream the same prompt again and print only the actual text content.
# The `end=""` argument keeps the output in a single flowing block.
response = llm.stream("Tell me about Agentic AI and how it is different from traditional AI models.")
for chunk in response:
    print(chunk.content, end = "")
```

```python
# Switch to a Groq-hosted model to compare streaming behavior with a cloud model.
# This requires the appropriate Groq API key to be available in the environment.
from langchain_groq import ChatGroq
llm  = ChatGroq(model="qwen/qwen3.8-27b", streaming=True)
```

```python
# Stream a response from the Groq model and print the content as it arrives.
response = llm.stream("Tell me about Agentic AI and how it is different from traditional AI models.")
for chunk in response:
    print(chunk.content, end = "")
```
