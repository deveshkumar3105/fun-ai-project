```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model = "gemini-3.5-flash-lite")
```

```python
text = """
My name is Devesh Kumar. I am a software engineer with a passion for AI and machine learning. I have experience in developing web applications and working with various programming languages and frameworks. In my free time, I enjoy exploring new technologies and contributing to open-source projects. I am 28 years old and currently based in San Francisco, California. I hold a Bachelor's degree in Computer Science from XYZ University. My interests include natural language processing, computer vision, and data science. I am always eager to learn and take on new challenges in the field of AI. My email address is devesh.kumar@example.com.
"""
```

```python
res = llm.invoke("Find my name, age, location, and email address in the following text and return them in a structured format: " + text)
```

```python
print(type(res.content[0]['text']))  # Print the type of the AI's response
```

### Structured Output Example

```python
from pydantic import BaseModel, Field

class ResposeStructure(BaseModel):
    name: str = Field(description="The name of the User.")
    age: int = Field(description="The age of the User.")
    location: str = Field(description="The location of the User.")
    email: str = Field(description="The email address of the User.")

structured_output = llm.with_structured_output(ResposeStructure)
```

```python
res = structured_output.invoke("Find my name, age, location, and email address in the following text and return them in a structured format: " + text)
```

```python
res.name
res.age
res.location
res.email
```

```python
print(type(res))  # Print the type of the structured output
```

```python
res.model_dump()  # Print the structured output as a dictionary
```

```python
class Movie(BaseModel):
    title: str = Field(description="The title of the movie.")
    year: int = Field(description="The release year of the movie.")
    genre: str = Field(description="The genre of the movie.")
    director: str = Field(description="The director of the movie.")

class AllMovies(BaseModel):
    movies: list[Movie] = Field(description="A list of movies.")

movie_structured_output = llm.with_structured_output(Movie)
```

```python
movie_structured_output.invoke("Give me a science fiction movie.")
```

```python
movies_llm = llm.with_structured_output(AllMovies)
res = movies_llm.invoke("Give me a list of 3 science fiction movies.")
```
