```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langgraph.checkpoint.memory import MemorySaver
```

```python
llm = ChatGroq(model="qwen/qwen3.8-27b")
search = GoogleSerperAPIWrapper()
```

```python
search.run("Who won IPL 2026")
```

```python
agent = create_agent(
    model=llm,
    tools=[search.run],
    system_prompt="You are an agent and can search Google for responses",
    checkpointer=MemorySaver()
)
```

```python
question = "What is the current Dollar rate vs All currencies ?"
```

```python
# Here the thread id is the one that determines how the results are saved.
# Change the thread id to something else and the chat looses track of the conversation.
response = agent.invoke(
    {"messages":[{"role":"user","content":question}]},
    {"configurable":{"thread_id":"devesh"}},
    )
```

```python
response["messages"]
```

```python
response["messages"][-1].content
```

```python
from IPython.display import Markdown, display
final_text = response["messages"][-1].content
display(Markdown(final_text))
```
