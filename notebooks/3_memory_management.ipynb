## Manual memory management

```python
from dotenv import load_dotenv
load_dotenv()  # Load environment variables from .env file
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model = "gemini-3.5-flash-lite")
```

```python
prompts = [
    {"role": "user", "content": "Hello, My name is Devesh Kumar."},
    {"role": "ai", "content": "Hey Devesh! How can I assist you today?"},
    {"role": "user", "content": "Tell me who I am."}
]

res = llm.invoke(prompts)
```

```python
print(res.content[0]['text'])  # Print the AI's response
```

### Dynamic Memory Management - History Retention

```python
history = []
while True:
    query = input("You: ")

    if query.lower() in ["exit", "quit"]:
        print("AI : Goodbye 👋🏼")
        break

    history.append({"role": "user", "content": query})
    print("User: ", query, "\n")
    res = llm.invoke(history)
    history.append({"role": "ai", "content": res.content[0]['text']})
    print(f"AI: {res.content[0]['text']}\n")
```
