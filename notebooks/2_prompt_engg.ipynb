### Prompt Engg - Dynamic Prompts

```python
# Load Env
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
out = StrOutputParser()
```

```python
prompts = ChatPromptTemplate.from_messages([
    {"role": "system", "content": "You are a translator, translate the text into {language}"},
    {"role": "user", "content": "{query}"}
])
```

```python
print(prompts)
```

```python
promptsFilled = prompts.format_messages(language="French", query="Hello, how are you?")
res = llm.invoke(promptsFilled)
print(res.content[0]['text'])
```

```python
res = llm.invoke(prompts.invoke({"language": "Hindi", "query": "Hello, how are you?"}))
print(res.content[0]['text'])
```

```python
# Custom Chain parser update function
def to_upper(result:str):
    return result.upper()
```

```python
chains = prompts | llm | out | to_upper
```

```python
res = chains.invoke({"language": "Spanish", "query": "Hello, how are you?"})
```
